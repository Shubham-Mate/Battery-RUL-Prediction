```python
def drop_outlier_discharge_cycles(df,bins):
    index = []
    discharge_capacities = df['Discharge_Capacity(Ah)']
    range_ = np.arange(1,df.shape[0],bins)
    for i in range_[:-1]:
        array_lim = discharge_capacities[i:i+bins]
        sigma = np.std(array_lim)
        mean = np.median(array_lim)
        th_max,th_min = mean + sigma*1.2, mean - sigma*1.2
        idx = np.where((array_lim < th_max) & (array_lim > th_min))
        idx = idx[0] + i
        index.extend(list(idx))
    return df.iloc[index, :]
```

## Custom Dataset

```python
dfs = {name: pd.read_csv(path) for name, path in zip(CustomDatasetConstants.BATTERY_NAMES.value, CustomDatasetConstants.BATTERY_RAW_DATASET_PATHS.value)}

dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]] = dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]][dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]]['cap']!=0]
dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]] = dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]][dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]]['cap']!=0]
```

```python
dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]] = dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]].iloc[10000:, :]
```

```python
dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]] = dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]].loc[(dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]]["state"] != 0)]
dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]] = dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]].reset_index(drop=True)
#dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]] = dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]].iloc[10000:, :]
dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]] = dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]][~((dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]]["state"] == 0))]
dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]] = dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]].reset_index(drop=True)
```

```python
plt.figure(figsize=(10, 5))  # Set figure size
plt.plot(dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]]['cap'], linestyle='-', color='b', label='Cap')  # Line plot
plt.xlabel('Index')  # X-axis label
plt.ylabel('Cap Value')  # Y-axis label
plt.title('Line Plot of Cap Column')  # Title
plt.legend()  # Show legend
plt.grid(True)  # Add grid
plt.show() 
```

```python
dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]][dfs[CustomDatasetConstants.BATTERY_NAMES.value[0]]['state']==0]
```

```python
dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]]['cap_diff'] = dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]]['cap'].diff().abs()
dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]]['cap_diff'].describe()
```

```python
dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]]['group'] = (dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]]['cap_diff'] > 10).cumsum()
dfs_separated = {group: group_df for group, group_df in dfs[CustomDatasetConstants.BATTERY_NAMES.value[1]].groupby("group")}
dfs_separated
```

```python
dfs_separated[1]
```

```python
# Number of subplots = number of segments
num_segments = len(dfs_separated)

fig, axes = plt.subplots(num_segments, 1, figsize=(10, 4 * num_segments))

# If there's only one segment, axes won't be an iterable, so we handle it
if num_segments == 1:
    axes = [axes]

for (group, df_subset), ax in zip(dfs_separated.items(), axes):
    ax.plot(df_subset.index, df_subset["cap"], linestyle="-", label=f"Segment {group}")
    ax.set_ylabel("Capacity (Ah)")
    ax.legend()
    ax.grid()
    
    # Set custom x-ticks for each subplot
    ax.set_xticks(df_subset.index[::max(1, len(df_subset) // 10)])  # Show ~10 ticks per plot
    ax.set_xticklabels(df_subset.index[::max(1, len(df_subset) // 10)], rotation=45)

plt.xlabel("Index")
#plt.title("Capacity Curves for Each Separated Segment")
plt.tight_layout()
plt.show()
```
