```python
def drop_outlier_discharge_cycles(df,bins):
    index = []
    discharge_capacities = df['Discharge_Capacity(Ah)']
    range_ = np.arange(1,df.shape[0],bins)
    for i in range_[:-1]:
        array_lim = discharge_capacities[i:i+bins]
        sigma = np.std(array_lim)
        mean = np.median(array_lim)
        th_max,th_min = mean + sigma*1.2, mean - sigma*1.2
        idx = np.where((array_lim < th_max) & (array_lim > th_min))
        idx = idx[0] + i
        index.extend(list(idx))
    return df.iloc[index, :]
```

```python
capacities_out = preprocessing_functions.drop_outlier_discharge_cycles(capacities, 10)
capacities_out['Cycle_Index'] = np.arange(1, capacities_out.shape[0]+1)
capacities_out
```
